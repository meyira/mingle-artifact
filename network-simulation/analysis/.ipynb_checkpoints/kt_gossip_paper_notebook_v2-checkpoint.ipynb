## 0) Environment setup and simulator entrypoint
This cell:
- finds your repo root
- prefers `.venv` python if present
- detects simulator entrypoint
- discovers supported CLI flags from `--help` (so notebook never passes unknown args)


```python

import os, sys, re, json, subprocess
from pathlib import Path

# --- repo root discovery ---
HERE = Path.cwd()
candidates = [HERE] + list(HERE.parents)
REPO_ROOT = None
for p in candidates:
    if (p / "sim").exists():
        REPO_ROOT = p
        break
if REPO_ROOT is None:
    REPO_ROOT = HERE
print("Repo root:", REPO_ROOT)

# --- prefer root .venv python ---
venv_py_win = REPO_ROOT / ".venv" / "Scripts" / "python.exe"
venv_py_unix = REPO_ROOT / ".venv" / "bin" / "python"
if venv_py_win.exists():
    PYTHON_EXE = str(venv_py_win)
elif venv_py_unix.exists():
    PYTHON_EXE = str(venv_py_unix)
else:
    PYTHON_EXE = sys.executable
print("Python exe:", PYTHON_EXE)

def run_cmd(cmd, cwd=REPO_ROOT, check=False):
    print(">", " ".join(map(str, cmd)))
    p = subprocess.run(list(map(str, cmd)), cwd=cwd, capture_output=True, text=True)
    if p.stdout.strip():
        print(p.stdout)
    if p.stderr.strip():
        print("STDERR:\n", p.stderr)
    if check and p.returncode != 0:
        raise RuntimeError(f"Command failed rc={p.returncode}")
    return p

# --- detect entrypoint ---
def detect_entrypoint():
    p = subprocess.run([PYTHON_EXE, "-m", "sim.run", "--help"], cwd=REPO_ROOT, capture_output=True, text=True)
    if p.returncode == 0:
        return ["-m", "sim.run"]
    run_py = REPO_ROOT / "sim" / "run.py"
    if run_py.exists():
        p2 = subprocess.run([PYTHON_EXE, str(run_py), "--help"], cwd=REPO_ROOT, capture_output=True, text=True)
        if p2.returncode == 0:
            return [str(run_py)]
    raise RuntimeError("Could not detect simulator entrypoint. Try `python -m sim.run --help` in repo root.")

ENTRY = detect_entrypoint()
print("Detected ENTRY:", ENTRY)

# --- discover supported flags from --help ---
help_p = run_cmd([PYTHON_EXE] + ENTRY + ["--help"])
HELP_TEXT = (help_p.stdout or "") + "\n" + (help_p.stderr or "")
SUPPORTED_FLAGS = set(re.findall(r"(--[a-zA-Z0-9_]+)", HELP_TEXT))
print("Supported flags:", len(SUPPORTED_FLAGS))
```

## 2) Runner helpers (run_sim)
This cell provides a robust `run_sim(params, outdir)` wrapper and filters args through supported flags.

```python

import numpy as np

def params_to_args(params: dict):
    args = []
    dropped = []
    for k, v in params.items():
        flag = f"--{k}"
        if flag not in SUPPORTED_FLAGS:
            dropped.append(flag)
            continue
        if v is None:
            continue
        if isinstance(v, bool):
            if v:
                args.append(flag)
        else:
            args.extend([flag, str(v)])
    if dropped:
        print("Note: dropped unsupported flags:", ", ".join(sorted(set(dropped))))
    return args

def run_sim(params: dict, outdir: str):
    outpath = Path(outdir)
    if not outpath.is_absolute():
        outpath = REPO_ROOT / outpath
    outpath.parent.mkdir(parents=True, exist_ok=True)

    p = dict(params)
    p["outdir"] = str(outpath)

    cmd = [PYTHON_EXE] + ENTRY + params_to_args(p)
    rc = run_cmd(cmd).returncode
    if rc != 0:
        raise RuntimeError(f"Simulation failed (rc={rc}). See stderr above.")
    return outpath
```

## 3) Load outputs + paper-grade derived metrics
This cell:
- loads `summary.json`, `timeline.csv`, `events.csv`
- computes detection using proof counters (`strict` vs `soft`)
- computes **bytes per send** and **bytes per delivery** (paper-grade)


```python

import pandas as pd

def load_run(outdir: Path):
    outdir = Path(outdir)
    summary_path  = outdir / "summary.json"
    timeline_path = outdir / "timeline.csv"
    events_path   = outdir / "events.csv"

    summary  = json.loads(summary_path.read_text(encoding="utf-8")) if summary_path.exists() else {}
    timeline = pd.read_csv(timeline_path) if timeline_path.exists() else pd.DataFrame()
    events   = pd.read_csv(events_path) if events_path.exists() else pd.DataFrame()
    return summary, timeline, events

def derived_metrics(summary: dict, fallback: dict, detection_policy: str = "strict"):
    # knobs
    users      = float(summary.get("users", fallback.get("users", np.nan)))
    hours      = float(summary.get("hours", fallback.get("hours", np.nan)))
    dt_min     = float(summary.get("dt_min", fallback.get("dt_min", 5)))
    fork_hour  = float(summary.get("fork_hour", fallback.get("fork_hour", 0)))
    fork_frac  = float(summary.get("fork_frac", fallback.get("fork_frac", np.nan)))
    p_gossip   = float(summary.get("p_gossip", fallback.get("p_gossip", np.nan)))
    adoption   = float(summary.get("adoption", fallback.get("adoption", np.nan)))
    server_mode = summary.get("server_mode", fallback.get("server_mode", None))
    mode        = summary.get("mode", fallback.get("mode", None))
    seed        = int(summary.get("seed", fallback.get("seed", -1)))

    # proof counters
    pr_ok   = int(summary.get("proof_ok", 0))
    pr_inc  = int(summary.get("proof_inconsistent", 0))
    pr_with = int(summary.get("proof_withhold", 0))
    pr_to   = int(summary.get("proof_timeout", 0))

    crosscut_deliveries = int(summary.get("crosscut_deliveries", 0))
    crosscut_dropped    = int(summary.get("crosscut_dropped", 0))

    # detection definition
    if detection_policy == "soft":
        detected = (pr_inc + pr_with + pr_to) > 0
    else:
        detected = pr_inc > 0

    # latency
    first_tick = summary.get("first_detection_tick", None)
    t_fork_min = fork_hour * 60.0
    latency_min = np.nan
    if detected and first_tick not in (None, ""):
        try:
            t_detect_min = float(first_tick) * dt_min
            latency_min = max(0.0, t_detect_min - t_fork_min)
        except Exception:
            pass

    # overhead counters
    total_gossip_bytes = summary.get("total_gossip_bytes", np.nan)
    try: total_gossip_bytes = float(total_gossip_bytes)
    except: total_gossip_bytes = np.nan

    # NEW: paper-grade denominators if present
    total_sends = summary.get("total_sends", np.nan)
    total_deliv = summary.get("total_deliveries", summary.get("total_messages", np.nan))
    try: total_sends = float(total_sends)
    except: total_sends = np.nan
    try: total_deliv = float(total_deliv)
    except: total_deliv = np.nan

    bytes_per_send = (total_gossip_bytes / total_sends) if np.isfinite(total_gossip_bytes) and total_sends > 0 else np.nan
    bytes_per_delivery = (total_gossip_bytes / total_deliv) if np.isfinite(total_gossip_bytes) and total_deliv > 0 else np.nan

    # auxiliary rates
    crosscut_total = crosscut_deliveries + crosscut_dropped
    crosscut_drop_rate_obs = (crosscut_dropped / crosscut_total) if crosscut_total > 0 else np.nan

    proof_total = pr_ok + pr_inc + pr_with + pr_to
    pr_inc_rate = (pr_inc / proof_total) if proof_total > 0 else np.nan

    return dict(
        server_mode=server_mode, mode=mode, seed=seed,
        users=users, hours=hours, dt_min=dt_min, fork_hour=fork_hour, fork_frac=fork_frac,
        p_gossip=p_gossip, adoption=adoption,
        detected=detected, first_detection_tick=float(first_tick) if first_tick not in (None, "") else np.nan, latency_min=latency_min,

        total_gossip_bytes=total_gossip_bytes,
        total_sends=total_sends, total_deliveries=total_deliv,
        bytes_per_send=bytes_per_send,
        bytes_per_delivery=bytes_per_delivery,

        crosscut_deliveries=crosscut_deliveries,
        crosscut_dropped=crosscut_dropped,
        crosscut_drop_rate_obs=crosscut_drop_rate_obs,

        proof_ok=pr_ok,
        proof_inconsistent=pr_inc,
        proof_withhold=pr_with,
        proof_timeout=pr_to,
        proof_total=proof_total,
        proof_inconsistent_rate=pr_inc_rate,
    )
```

## 4) Sanity probes (A/B + harder targeted probes)
These probes confirm the core logic before any sweeps:
- Probe A: no gossip + no adopters ⇒ no detection
- Probe B: sticky vs non-sticky under low gossip
- Hard probes: reduce cross-boundary mixing (if supported) and/or remove groups


```python

def run_case(name, overrides, out_root="out_sanity_v2"):
    params = dict(BASE)

    # Avoid conflicting boolean flags
    params.pop("sticky_victims", None)
    params.pop("no_sticky_victims", None)

    params.update(overrides)

    outdir = REPO_ROOT / out_root / name
    outdir.mkdir(parents=True, exist_ok=True)

    # Force rerun by deleting old summary if present
    s = outdir / "summary.json"
    if s.exists():
        s.unlink()

    run_sim(params, outdir=str(outdir))
    summary, timeline, events = load_run(outdir)

    pr_inc = int(summary.get("proof_inconsistent", 0))
    pr_with = int(summary.get("proof_withhold", 0))
    pr_to = int(summary.get("proof_timeout", 0))
    detected_strict = pr_inc > 0
    detected_soft = (pr_inc + pr_with + pr_to) > 0

    return dict(
        name=name,
        p_gossip=float(summary.get("p_gossip", params.get("p_gossip", np.nan))),
        adoption=float(summary.get("adoption", params.get("adoption", np.nan))),
        fork_frac=float(summary.get("fork_frac", params.get("fork_frac", np.nan))),
        p_contact_cross=float(summary.get("p_contact_cross", params.get("p_contact_cross", np.nan))),
        p_group_cross=float(summary.get("p_group_cross", params.get("p_group_cross", np.nan))),
        groups=float(summary.get("groups", params.get("groups", np.nan))),
        msg_rate_group=float(summary.get("msg_rate_group", params.get("msg_rate_group", np.nan))),
        crosscut_deliveries=int(summary.get("crosscut_deliveries", 0)),
        crosscut_dropped=int(summary.get("crosscut_dropped", 0)),
        proof_inconsistent=pr_inc,
        first_detection_tick=summary.get("first_detection_tick", np.nan),
        detected_strict=detected_strict,
        detected_soft=detected_soft,
    )

# Probe A
A = run_case("ProbeA_sticky_no_gossip",
             dict(server_mode="permanent_fork", mode="sth_proof",
                  sticky_victims=True, fork_frac=0.1,
                  p_gossip=0.0, adoption=0.0))

# Probe B
B1 = run_case("ProbeB_nonsticky_low",
              dict(server_mode="permanent_fork", mode="sth_proof",
                   no_sticky_victims=True, fork_frac=0.1,
                   p_gossip=0.05, adoption=0.2))

B2 = run_case("ProbeB_sticky_low",
              dict(server_mode="permanent_fork", mode="sth_proof",
                   sticky_victims=True, fork_frac=0.1,
                   p_gossip=0.05, adoption=0.2))

df_probe = pd.DataFrame([A, B1, B2])
df_probe
```

## 5) Sweep utilities (SELF-CONTAINED)
Defines `run_grid()` so you never hit `NameError: run_grid not defined` after kernel restarts.

Also uses short directory names to avoid Windows path-length problems.


```python

from itertools import product

def safe_name(v):
    if isinstance(v, float):
        return f"{v:g}".replace(".", "p")
    return str(v).replace("/", "_")

def make_outdir_name(cfg, seed):
    sm = cfg.get("server_mode", "m")
    mode = cfg.get("mode", "x")
    pg = safe_name(cfg.get("p_gossip", 0))
    a  = safe_name(cfg.get("adoption", 0))
    f  = safe_name(cfg.get("fork_frac", 0))
    # optional: mixing knobs if present
    pc = cfg.get("p_contact_cross", None)
    pgc = cfg.get("p_group_cross", None)
    mix = ""
    if pc is not None or pgc is not None:
        mix = f"_pc{safe_name(pc if pc is not None else 1.0)}_pg{safe_name(pgc if pgc is not None else 1.0)}"
    return f"{sm}_{mode}_pg{pg}_a{a}_f{f}{mix}_s{seed}"

def has_summary(outdir: Path):
    outdir.mkdir(parents=True, exist_ok=True)
    return (outdir / "summary.json").exists()

def run_grid(base: dict, sweep: dict, replicates: int, seed_base: int, outroot: Path, detection_policy="strict"):
    outroot = Path(outroot)
    outroot.mkdir(parents=True, exist_ok=True)

    results = []
    grid_keys = list(sweep.keys())
    total = int(np.prod([len(sweep[k]) for k in grid_keys]) * replicates)
    done = 0
    skipped = 0

    for values in product(*[sweep[k] for k in grid_keys]):
        cfg = dict(base)
        cfg.update({k: v for k, v in zip(grid_keys, values)})

        for r in range(replicates):
            seed = seed_base + r
            cfg_run = dict(cfg)
            cfg_run["seed"] = seed

            outdir = outroot / make_outdir_name(cfg_run, seed)

            if not has_summary(outdir):
                run_sim(cfg_run, outdir=str(outdir))
            else:
                skipped += 1

            summary, timeline, events = load_run(outdir)
            dm = derived_metrics(summary, fallback=cfg_run, detection_policy=detection_policy)
            dm["run_dir"] = str(outdir)
            results.append(dm)

            done += 1
            if done % 25 == 0:
                print(f"{done}/{total} done (skipped={skipped})")

    return pd.DataFrame(results)

def aggregate(df_raw: pd.DataFrame):
    df = df_raw.copy()

    # numeric coercion
    for c in ["p_gossip","adoption","fork_frac","latency_min","bytes_per_send","bytes_per_delivery",
              "p_contact_cross","p_group_cross"]:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")

    # group-by columns (extend if present)
    gcols = ["server_mode","mode","p_gossip","adoption","fork_frac"]
    for extra in ["p_contact_cross","p_group_cross","groups","msg_rate_group"]:
        if extra in df.columns:
            gcols.append(extra)

    agg = df.groupby(gcols, dropna=False).agg(
        reps=("run_dir","count"),
        det_rate=("detected","mean"),
        lat_med=("latency_min","median"),
        lat_mean=("latency_min","mean"),
        bps_mean=("bytes_per_send","mean"),
        bpd_mean=("bytes_per_delivery","mean"),
        crosscuts=("crosscut_deliveries","mean"),
    ).reset_index()

    return agg
```

## 6) Targeted victims scenario (sticky + reduced mixing)
Your earlier sweep saturated because large-scale groups cause immediate crosscuts. For a meaningful targeted scenario, start by:
- `sticky_victims=True`
- smaller victim cohort (`fork_frac`)
- reduce cross-boundary mixing (if your patched code supports `p_contact_cross` / `p_group_cross`)
- optionally disable groups for a pure 1:1 substrate

Below is a *small* targeted sweep. Adjust `HARD` knobs until you see `det_rate<1` and `lat_med>0` in some corner.


```python

# Use a fresh sweep folder so nothing is skipped
SWEEP_ROOT = REPO_ROOT / "out_nb_v3"
SWEEP_ROOT.mkdir(parents=True, exist_ok=True)

OUT_TARG = SWEEP_ROOT / "targeted_sticky_small_1"
OUT_TARG.mkdir(parents=True, exist_ok=True)

BASE_TARGETED = dict(BASE)
BASE_TARGETED.update(
    server_mode="permanent_fork",
    mode="sth_proof",
    sticky_victims=True,
    fork_frac=0.05,  # smaller victim set is harder
)

# HARDENING knobs (uncomment if supported)
# BASE_TARGETED.update(p_contact_cross=0.02, p_group_cross=0.02)

# Optional: reduce group mixing radically
# BASE_TARGETED.update(groups=0, msg_rate_group=0.0)

SWEEP_TARGETED_SMALL = dict(
    p_gossip=[0.01, 0.05, 0.2],
    adoption=[0.1, 0.3, 0.6, 1.0],
)

df_raw_targ = run_grid(BASE_TARGETED, SWEEP_TARGETED_SMALL,
                       replicates=10, seed_base=8000, outroot=OUT_TARG,
                       detection_policy="strict")

df_raw_targ.to_csv(OUT_TARG / "all_summaries_raw.csv", index=False)
df_agg_targ = aggregate(df_raw_targ)
df_agg_targ.to_csv(OUT_TARG / "all_summaries_agg.csv", index=False)

print("Wrote:", OUT_TARG)
df_agg_targ.sort_values(["adoption","p_gossip"])

OUT_TARG2 = SWEEP_ROOT / "targeted_sticky_lowmix"
OUT_TARG2.mkdir(parents=True, exist_ok=True)
BASE_TARGETED2 = dict(BASE_TARGETED)
BASE_TARGETED2.update(p_contact_cross=0.02, p_group_cross=0.02)
df_raw_targ2 = run_grid(BASE_TARGETED2, SWEEP_TARGETED_SMALL,
                       replicates=10, seed_base=9000, outroot=OUT_TARG2,
                       detection_policy="strict")
df_agg_targ2 = aggregate(df_raw_targ2)
df_agg_targ2.sort_values(["adoption","p_gossip"])
```

## 7) Paper plots for a sweep (heatmaps + tradeoff)
Use this cell after a sweep to produce:
- detection-rate heatmap
- median-latency heatmap
- overhead-vs-latency scatter (bytes per delivery)


```python

import matplotlib.pyplot as plt
import matplotlib as mpl
mpl.rcParams.update({"font.size": 9, "pdf.fonttype": 42, "ps.fonttype": 42})

def heatmap(df_agg, value_col, title, outprefix, xcol="p_gossip", ycol="adoption"):
    hm = df_agg.copy()
    xs = sorted(hm[xcol].dropna().unique())
    ys = sorted(hm[ycol].dropna().unique())
    X = {x:i for i,x in enumerate(xs)}
    Y = {y:i for i,y in enumerate(ys)}
    Z = np.full((len(ys), len(xs)), np.nan)
    for _, r in hm.iterrows():
        Z[Y[r[ycol]], X[r[xcol]]] = r[value_col]
    plt.figure()
    plt.imshow(Z, aspect="auto", origin="lower")
    plt.xticks(range(len(xs)), [f"{x:g}" for x in xs], rotation=45)
    plt.yticks(range(len(ys)), [f"{y:g}" for y in ys])
    plt.xlabel(r"$p_{gossip}$")
    plt.ylabel(r"adoption $a$")
    plt.title(title)
    plt.colorbar()
    plt.tight_layout()
    plt.savefig(outprefix.with_suffix(".pdf"))
    plt.savefig(outprefix.with_suffix(".png"), dpi=300)
    plt.show()

# Use targeted sweep by default (change df_agg_* and OUT_* if you run other sweeps)
heatmap(df_agg_targ, "det_rate", "Detection rate (targeted sticky)", OUT_TARG / "fig_heat_det_rate")
heatmap(df_agg_targ, "lat_med", "Median detection latency (min)", OUT_TARG / "fig_heat_latency")

# Tradeoff: overhead vs latency using bytes per delivery
df = df_agg_targ.copy()
df = df[np.isfinite(df["bpd_mean"]) & np.isfinite(df["lat_med"])]

plt.figure()
plt.scatter(df["bpd_mean"], df["lat_med"])
plt.xlabel("Mean gossip bytes per delivery")
plt.ylabel("Median detection latency (min)")
plt.title("Overhead vs detection latency (targeted)")
plt.tight_layout()
plt.savefig(OUT_TARG / "fig_tradeoff.pdf")
plt.savefig(OUT_TARG / "fig_tradeoff.png", dpi=300)
plt.show()
```

## 8) Export LaTeX table (USENIX two-column safe)
Exports `table_top12.tex` sorted by: detection rate ↓, latency ↑ (lower better), overhead ↓.
Uses bytes per delivery.


```python

tbl = df_agg_targ.copy().sort_values(["det_rate","lat_med","bpd_mean"], ascending=[False, True, True])
tbl = tbl[["p_gossip","adoption","reps","det_rate","lat_med","bpd_mean","crosscuts"]].copy()
tbl = tbl.rename(columns={
    "p_gossip": r"$p_{gossip}$",
    "adoption": r"$a$",
    "reps": "reps",
    "det_rate": "Detect",
    "lat_med": "MedLat(min)",
    "bpd_mean": "Bytes/deliv",
    "crosscuts": "Crosscuts",
})

latex = tbl.head(12).to_latex(index=False, escape=False, float_format="%.3g")
out_tex = OUT_TARG / "table_top12.tex"
out_tex.write_text(
    "\\begin{table}[t]\n\\centering\\small\n"
    "\\caption{Top configurations from the targeted sticky sweep.}\n"
    "\\label{tab:top_targeted}\n"
    f"{latex}\n\\end{table}\n",
    encoding="utf-8"
)
print("Wrote:", out_tex)
tbl.head(12)
```
